## LEE CARPETA SIN SUBCARPETAS, Y LAS CREA X CLASES

```python
import os
import shutil
import random
from collections import defaultdict

# Diccionario de reglas de clasificación
CLASS_RULES = {
    "chat": ["chat", "Chat"],
    "email": ["email", "POP", "IMAP"],
    "voice": ["audio", "voipbuster", "voice"],
    "streaming": ["vimeo", "netflix", "spotify", "youtube", "video"],
    "filetransfer": ["sftp", "ftps", "files", "file", "filetransfer", "transfer", "ssl"],
    "p2p": ["bittorrent"],
    "browsing": ["browsing"]
}

# Directorio de entrada (donde se encuentran las fotos a chorro) y salida (donde se ordenarán por clase)
input_dir = r"C:\Users\migue\Desktop\Cositis\unistuff\TFG\mio\data\TESTING\zorder"
output_dir = r"C:\Users\migue\Desktop\Cositis\unistuff\TFG\mio\data\TESTING\nnigga"

# Límite por clase
MAX_PER_CLASS = 500_000

os.makedirs(output_dir, exist_ok=True)
class_counts = defaultdict(int)

all_files = [f for f in os.listdir(input_dir) if f.endswith(".png")]
random.shuffle(all_files)

for filename in all_files:
    file_path = os.path.join(input_dir, filename)

    label_found = False
    for label, keywords in CLASS_RULES.items():
        if any(keyword.lower() in filename.lower() for keyword in keywords):
            if class_counts[label] >= MAX_PER_CLASS:
                label_found = True
                break
            dest_dir = os.path.join(output_dir, label)
            os.makedirs(dest_dir, exist_ok=True)
            shutil.copy2(file_path, os.path.join(dest_dir, filename))
            class_counts[label] += 1
            label_found = True
            break

    if not label_found:
        if class_counts["unclassified"] >= MAX_PER_CLASS:
            continue
        dest_dir = os.path.join(output_dir, "unclassified")
        os.makedirs(dest_dir, exist_ok=True)
        shutil.copy2(file_path, os.path.join(dest_dir, filename))
        class_counts["unclassified"] += 1

print("✅ Copiado completado. Resumen:")
for label, count in class_counts.items():
    print(f"  - {label}: {count} imágenes copiadas")
```
